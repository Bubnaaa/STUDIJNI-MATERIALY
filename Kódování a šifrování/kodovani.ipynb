## Unicode
- Tabulka, která každému znaku přiřazuje unikátní číslo

### Ukládání do paměti
#### UTF-8
- Používá prefixové kódy
- 1-4 bajty
- První bit určuje, zda jde o začátek nového znaku nebo pokračování (Pokud se uprostřed přenosu ztratí bit, systém pozná kde začíná další znak) 

#### UTF-16
- 16 bitové jednotky
- U UTF-16 je důležité pořadí bajtů:
    1. Little Endian - Méně významný bajt
    2. Big Endian - Více významný bajt

```python
text = "Ahoj"

data_utf8 = text.encode('utf-8')
data_utf16le = text.encode('utf-16-le')

hex_utf8 = data_utf8.hex(' ')
hex_utf16le = data_utf16le.hex(' ')

print(f"UTF-8 hex:    {hex_utf8}")
print(f"UTF-16LE hex: {hex_utf16le}")
```

.encode() - Transformuje znaky na konkrétní sekvenci bajtů podle zvolené sady

.hex() - Vezme každý bajt a převede ho za 0-255 na dvojici znaků 0-F

## Šifrování a dešifrování pomocí bitového XOR
#### XOR
- Je vratný - pokud aplikujete stejný klíč na zašifrovaná data, získáte zpět původní zprávu
- např: M+K=C , C+K=M
- Podle Claudeho Shannona je šifra matematicky neprolomitelná, pokud je klíč stejně dlouhý jako šifra a je použit jen jednou

```python
def xor_cipher(data: bytes, key: bytes) -> bytes:
    return bytes([b ^ key[i % len(key)] for i, b in enumerate(data)])

original_data = "Tajná zpráva".encode('utf-8')
key = b"klic123"

encrypted = xor_cipher(original_data, key)
print(f"Zašifrováno (hex): {encrypted.hex(' ')}")

decrypted = xor_cipher(encrypted, key)
print(f"Dešifrováno: {decrypted.decode('utf-8')}")
```

- Operátor ^ v pythonu provádí bitový XOR
- Pokud je klíč kratší než zpráva, používáme operaci %, abychom se v klíči vraceli na začátek

## Změna kódování textového souboru
- Při převodu kódování (např. z UTF-8 na UTF-16) neprovádíme přímou transformaci bajtů

- Dekódování: Převod sekvence bajtů (např. UTF-8) zpět do abstraktního prostoru Unicode (Code Points)

- Rekódování: Mapa z abstraktního prostoru Unicode do nové cílové binární struktury (např. UTF-16)

```python
input_file = 'vstup.txt'
output_file = 'vystup_utf16.txt'

with open(input_file, 'r', encoding='utf-8') as f:
    obsah = f.read()

with open(output_file, 'w', encoding='utf-16-le') as f:
    f.write(obsah)
```

- Klíčem je parametr encoding ve funkci open
- Při čtení Python interně převede bajty z disku na Unicode řetězec
